```python
# Basic libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
# Preprocessing
from sklearn.preprocessing import StandardScaler

# Clustering
from sklearn.cluster import KMeans
from sklearn.cluster import AgglomerativeClustering

# Evaluation metrics
from sklearn.metrics import (
    silhouette_score,
    davies_bouldin_score,
    calinski_harabasz_score
)

# Dimensionality reduction
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE

# Dendrogram
from scipy.cluster.hierarchy import dendrogram, linkage

RANDOM_STATE = 42
```

**Load the Dataset**

```python
df_cluster = pd.read_csv("Crop_recommendation.csv")

print("Dataset Shape:", df_cluster.shape)
display(df_cluster)
```

**Dataset information**

```python
print("Dataset Information:")
df_cluster.info()
print("\nMissing Values:")
print(df_cluster.isnull().sum())
print("\nDuplicate Rows:")
print(df_cluster.duplicated().sum())
print("\nDescribe:")
print(df_cluster.describe())
df_cluster = df_cluster.drop_duplicates().reset_index(drop=True)
print("Shape after removing duplicates:", df_cluster.shape)
```

**Select Features for Clustering**

```python
cluster_features = [
    "N",
    "P",
    "K",
    "temperature",
    "humidity",
    "ph",
    "rainfall"
]

X_cluster = df_cluster[cluster_features]

print("Features used for clustering:")
display(X_cluster.head())
```

**Standardization**

```python
scaler_cluster = StandardScaler()
X_scaled = scaler_cluster.fit_transform(X_cluster)
print("Scaled Data Shape:", X_scaled.shape)
```

**Exploratory Correlation Heatmap**

```python
plt.figure(figsize=(9, 7))
sns.heatmap(
    X_cluster.corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)
plt.title("Correlation Heatmap of Clustering Features")
plt.tight_layout()
plt.show()
```

The heatmap indicates that most clustering features have weak or negligible linear correlations, suggesting good feature diversity. The notable exception is the strong positive correlation between phosphorus (P) and potassium (K), with a correlation of 0.74. Therefore, P and K contain some overlapping information, while the other environmental and soil features contribute relatively independent information to the clustering process.

**K-Means — Finding Optimal K**

```python
warnings.filterwarnings("ignore")
inertia_values = []

K_range = range(2, 11)

for k in K_range:
    kmeans = KMeans(
        n_clusters=k,
        random_state=RANDOM_STATE,
        n_init=10
    )
    
    kmeans.fit(X_scaled)
    inertia_values.append(kmeans.inertia_)
plt.figure(figsize=(8, 5))

plt.plot(
    K_range,
    inertia_values,
    marker="o"
)

plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.title("Elbow Method for Optimal K")
plt.xticks(K_range)
plt.grid(True)

plt.tight_layout()
plt.show()
```

The Elbow Method suggests that K = 5 is a suitable choice for the number of clusters. At this point, the reduction in inertia starts becoming less significant, indicating that increasing the number of clusters beyond 5 provides diminishing improvement in within-cluster compactness.

**Train K-Means**

```python
OPTIMAL_K = 5
kmeans_model = KMeans(
    n_clusters=OPTIMAL_K,
    random_state=RANDOM_STATE,
    n_init=10
)

kmeans_labels = kmeans_model.fit_predict(X_scaled)

df_cluster["KMeans_Cluster"] = kmeans_labels

print("K-Means cluster counts:")
print(df_cluster["KMeans_Cluster"].value_counts().sort_index())
```

**K-Means Evaluation**

```python
kmeans_silhouette = silhouette_score(
    X_scaled,
    kmeans_labels
)

kmeans_db = davies_bouldin_score(
    X_scaled,
    kmeans_labels
)

kmeans_ch = calinski_harabasz_score(
    X_scaled,
    kmeans_labels
)

print("K-Means Evaluation")
print("----------------------------")
print("Silhouette Score:", round(kmeans_silhouette, 4))
print("Davies-Bouldin Index:", round(kmeans_db, 4))
print("Calinski-Harabasz Index:", round(kmeans_ch, 4))
```

**PCA — K-Means Visualization**

```python
pca = PCA(
    n_components=2,
    random_state=RANDOM_STATE
)

X_pca = pca.fit_transform(X_scaled)

print("Explained Variance Ratio:")
print(pca.explained_variance_ratio_)

print("Total Explained Variance:",
      round(pca.explained_variance_ratio_.sum(), 4))
plt.figure(figsize=(9, 6))

scatter = plt.scatter(
    X_pca[:, 0],
    X_pca[:, 1],
    c=kmeans_labels,
    cmap="viridis",
    s=40
)

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("K-Means Clusters - PCA 2D Visualization")
plt.colorbar(scatter, label="Cluster")

plt.tight_layout()
plt.show()
```

The PCA-based 2D visualization of the K-Means results shows five distinct clusters with varying degrees of separation. Clusters 3 and 4 exhibit relatively clear separation from the remaining observations, whereas Clusters 0, 1, and 2 show substantial overlap in the two-dimensional PCA space. This indicates that some groups have distinct feature patterns, while others share similar characteristics across the principal components. The overlap may also indicate that additional dimensions are required to fully distinguish the clusters.

**Agglomerative Hierarchical Clustering**

```python
hierarchical_model = AgglomerativeClustering(
    n_clusters=OPTIMAL_K,
    linkage="ward"
)

hierarchical_labels = hierarchical_model.fit_predict(X_scaled)

df_cluster["Hierarchical_Cluster"] = hierarchical_labels

print("Hierarchical cluster counts:")
print(
    df_cluster["Hierarchical_Cluster"]
    .value_counts()
    .sort_index()
)
```

**Hierarchical Clustering Evaluation**

```python
hier_silhouette = silhouette_score(
    X_scaled,
    hierarchical_labels
)

hier_db = davies_bouldin_score(
    X_scaled,
    hierarchical_labels
)

hier_ch = calinski_harabasz_score(
    X_scaled,
    hierarchical_labels
)

print("Agglomerative Hierarchical Clustering")
print("---------------------------------------")
print("Silhouette Score:", round(hier_silhouette, 4))
print("Davies-Bouldin Index:", round(hier_db, 4))
print("Calinski-Harabasz Index:", round(hier_ch, 4))
```

**Dendrogram**

```python
sample_size = min(300, len(X_scaled))

X_dendrogram = X_scaled[:sample_size]

linkage_matrix = linkage(
    X_dendrogram,
    method="ward"
)
plt.figure(figsize=(14, 7))

dendrogram(
    linkage_matrix,
    truncate_mode="lastp",
    p=30,
    leaf_rotation=90,
    leaf_font_size=10
)

plt.title("Hierarchical Clustering Dendrogram")
plt.xlabel("Cluster / Sample Group")
plt.ylabel("Distance")

plt.tight_layout()
plt.show()
```

The hierarchical clustering dendrogram illustrates the formation of clusters based on the distance between observations. Several observations are merged at low distances, indicating strong similarity, while major groups are combined only at higher distances, indicating substantial differences between them. The dendrogram therefore reveals a clear hierarchical structure within the dataset and can be used to determine an appropriate number of clusters by selecting a suitable distance threshold.

**PCA Visualization for Hierarchical Clustering**

```python
plt.figure(figsize=(9, 6))

scatter = plt.scatter(
    X_pca[:, 0],
    X_pca[:, 1],
    c=hierarchical_labels,
    cmap="plasma",
    s=40
)

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("Hierarchical Clusters - PCA 2D Visualization")
plt.colorbar(scatter, label="Cluster")

plt.tight_layout()
plt.show()
```

The PCA visualization of the hierarchical clustering results reveals five clusters with different levels of separation. Cluster 4 is clearly separated from the remaining observations along Principal Component 1, indicating a distinct group. Cluster 3 also forms a relatively compact group at lower Principal Component 2 values, while Clusters 0, 1, and 2 exhibit considerable overlap. This suggests that some observations have similar characteristics, whereas other groups possess more distinct feature patterns.

**t-SNE Visualization**

```python
tsne = TSNE(
    n_components=2,
    random_state=RANDOM_STATE,
    perplexity=30
)

X_tsne = tsne.fit_transform(X_scaled)
plt.figure(figsize=(9, 6))

scatter = plt.scatter(
    X_tsne[:, 0],
    X_tsne[:, 1],
    c=kmeans_labels,
    cmap="viridis",
    s=40
)

plt.xlabel("t-SNE Component 1")
plt.ylabel("t-SNE Component 2")
plt.title("K-Means Clusters - t-SNE Visualization")
plt.colorbar(scatter, label="Cluster")

plt.tight_layout()
plt.show()
```

The t-SNE visualization of the K-Means clustering results reveals several compact and well-separated groups in the two-dimensional embedding. Compared with the PCA representation, t-SNE provides clearer visualization of local cluster structures and reveals multiple distinct subgroups. Although some overlap exists in the central regions, most clusters form relatively compact groups, indicating meaningful underlying patterns within the dataset.

**Final Clustering Comparison Table**

```python
clustering_results = pd.DataFrame({
    "Algorithm": [
        "K-Means",
        "Agglomerative Hierarchical"
    ],
    "Silhouette Score": [
        kmeans_silhouette,
        hier_silhouette
    ],
    "Davies-Bouldin Index": [
        kmeans_db,
        hier_db
    ],
    "Calinski-Harabasz Index": [
        kmeans_ch,
        hier_ch
    ]
})

clustering_results = clustering_results.round(4)

display(clustering_results)
```
